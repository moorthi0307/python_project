```python
original=pd.read_csv("amazon.csv")
amazon_sales["actual_price"]=original["actual_price"]
amazon_sales["actual_price"].head(10)
```

```python
product_rating=amazon_sales.groupby("product_id")["rating"].sum().head(5)
plt.figure(figsize=(15,6))
bars=plt.bar(product_rating.index,product_rating.values,color="orange")
plt.bar_label(bars)
plt.xlabel("product_id")
plt.ylabel("rating")
plt.title("product_id & rating")
plt.xticks(rotation=50)
plt.show()
```

```python
product_price=amazon_sales.groupby("product_id")["actual_price"].sum().head(5)
fig=plt.figure(figsize=(10,3))
ax=fig.add_subplot(1,1,1)
ax.plot(product_price,color="red",marker="*")
plt.xlabel("product_id")
plt.ylabel("actual_price")
plt.title("product_id vs actual_price")
plt.grid(True)
plt.show()
```

```python
product_discount=amazon_sales["discount_percentage"].value_counts().head()
plt.figure(figsize=(8,8))
colors = ["#eb4b4b", "#268cf1", "#1fe91f", "#eca548","#e0ce24"]
plt.pie(product_discount,labels=product_discount.index,autopct="%1.1f%%",startangle=90,colors=colors,explode=[0.05,0.06,0.04,0.07,0])
plt.title("product_discount_price")
plt.show()
```

```python
plt.figure(figsize=(15,6))
colors=["red","blue","green","orange","purple","pink","yellow","cyan","brown","black"]
plt.scatter(
amazon_sales["actual_price"].head(10),
amazon_sales["discounted_price"].head(10),c=colors,marker="o")
plt.xlabel("actual_price")
plt.ylabel('discounted_price')
plt.title('actual_price vs discounted_price')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(20,10))
sns.histplot(amazon_sales["actual_price"].dropna().head(30),bins=10,color="green",fill=False,edgecolor="black",kde=True)
plt.xlabel("actual_price")
plt.ylabel("Number of products")
plt.title("actual_price vs Number of products")
plt.show()
```

```python
plt.figure(figsize=(10,4))
ax=sns.barplot(
x="product_id",
y="rating_count",
data=amazon_sales.head(10),
hue="product_id",
palette="viridis",
legend=False)
ax.bar_label
plt.xlabel("product_id")
plt.ylabel("rating_count")
plt.title("rating_count vs product_id")
plt.xticks(rotation=50)
plt.show()
```

```python
plt.figure(figsize=(12,6))
ax=sns.boxplot(
y="actual_price",
x="discount_percentage",
data=amazon_sales.head(10),color="purple")
plt.ylabel("actual_price",fontsize=12)
plt.xlabel("discount_percentage",fontsize=12)
plt.title("actual_price vs discount_percentage",fontsize=16)
plt.grid(axis="y",linestyle="--",alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

ax = sns.countplot(
    data=amazon_sales,
    x="rating",
    hue="rating"
)

plt.xlabel("Rating", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.title("Rating Distribution", fontsize=16)

for container in ax.containers:
    ax.bar_label(container, padding=3)

plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))

top10_category = amazon_sales["category"].value_counts().head(10)

ax = sns.barplot(
    x=top10_category.values,
    y=top10_category.index,
    color="green"
)

plt.xlabel("Number of Products", fontsize=12)
plt.ylabel("Category", fontsize=12)
plt.title("Top 10 Categories by Product Count", fontsize=16)

for container in ax.containers:
    ax.bar_label(container, padding=3)

plt.grid(axis="x", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))

ax = sns.violinplot(
    data=amazon_sales.head(20),
    x="rating",
    y="actual_price",
    color="green"
)

plt.xlabel("Rating", fontsize=12)
plt.ylabel("Actual Price", fontsize=12)
plt.title("Rating vs Actual Price", fontsize=16)

plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16,12))

# 1. Histogram
plt.subplot(2,2,1)
sns.histplot(
    data=amazon_sales.head(10),
    x="actual_price",
    bins=10,
   hue="actual_price",
    kde=True
)
plt.title("Actual Price Distribution")

# 2. Scatter Plot
plt.subplot(2,2,2)
sns.scatterplot(
    data=amazon_sales.head(10),
    x="actual_price",
    y="discounted_price",
    color="blue",
    s=100
)
plt.title("Actual Price vs Discounted Price")

# 3. Countplot
plt.subplot(2,2,3)
sns.countplot(
    data=amazon_sales.head(10),
    x="rating",
    color="orange"
)
plt.title("Rating Distribution")

# 4. Violin Plot
plt.subplot(2,2,4)
sns.violinplot(
    data=amazon_sales.head(10),
    x="rating",
    y="actual_price",
    color="purple"
)
plt.title("Rating vs Actual Price")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16,12))

# 1. Pie Chart
plt.subplot(2,2,2)

category_count = amazon_sales["category"].value_counts().head(6)

plt.pie(
    category_count.values,
    labels=category_count.index,
    autopct="%1.1f%%",
    startangle=90,
    )

plt.title("Category Distribution")


# 2. Bar Chart
plt.subplot(2,2,1)

top_category = amazon_sales["category"].value_counts().head(10)

ax = sns.barplot(
    x=top_category.values,
    y=top_category.index,
    color="green"
)

plt.title("Top 10 Categories")
plt.xlabel("Product Count")
plt.ylabel("Category")

for container in ax.containers:
    ax.bar_label(container)


# 3. Line Chart
plt.subplot(2,2,3)

price_data = amazon_sales["actual_price"].head(10)

plt.plot(
    range(1, len(price_data)+1),
    price_data,
    marker="o"
)

plt.title("Actual Price Trend")
plt.xlabel("Product Number")
plt.ylabel("Actual Price")


# 4. Boxplot
plt.subplot(2,2,4)

sns.boxplot(
    data=amazon_sales.head(10),
    x="rating",
    y="discount_percentage",
    color="purple"
)

plt.title("Discount Percentage by Rating")
plt.xlabel("Rating")
plt.ylabel("Discount Percentage")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(18,12))

# 1. Strip Plot
plt.subplot(2,3,1)
sns.stripplot(
    data=amazon_sales.head(20),
    x="rating",
    y="actual_price",
    color="green"
)
plt.title("Rating vs Actual Price")

# 2. Swarm Plot
plt.subplot(2,3,2)
sns.swarmplot(
    data=amazon_sales.head(20),
    x="rating",
    y="discount_percentage",
    color="blue"
)
plt.title("Rating vs Discount")

# 3. ECDF Plot
plt.subplot(2,3,3)
sns.ecdfplot(
    data=amazon_sales.head(20),
    x="actual_price",
    color="purple"
)
plt.title("Actual Price ECDF")

# 4. Point Plot
plt.subplot(2,3,4)
sns.pointplot(
    data=amazon_sales.head(20),
    x="rating",
    y="discount_percentage",
    color="orange"
)
plt.title("Rating vs Discount Percentage")


# 5. Boxen Plot

plt.subplot(2,3,5)

sns.boxenplot(
    data=amazon_sales.head(20),
    x="rating",
    y="actual_price",
    color="green"
)

plt.title("Rating vs Actual Price")
plt.xlabel("Rating")
plt.ylabel("Actual Price")

# 6. Heat Map
plt.subplot(2,3,6)
corr = amazon_sales.select_dtypes(include="number").corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="Greens",
    fmt=".2f"
)

plt.title("Amazon Sales Correlation Heatmap", fontsize=16)

plt.title("Actual Price Density")
plt.tight_layout()
plt.show()
```
